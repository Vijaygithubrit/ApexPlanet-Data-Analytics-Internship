# 9. Outlier Detection and Investigation

## Objective

The objective of this step is to identify statistically unusual
numerical observations in the dataset.

We will:

- Use the Interquartile Range (IQR) method.
- Identify potential outliers in numerical columns.
- Count the number of potential outliers.
- Inspect the identified records.

Potential outliers will not be removed automatically.
A separate cleaning decision will be made later.

```python
# Define the numerical columns for outlier analysis.

outlier_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

print("Columns selected for outlier analysis:")
print(outlier_columns)
```

```python
# Calculate the first quartile (Q1), third quartile (Q3),
# and Interquartile Range (IQR) for each numerical column.

outlier_summary = []

for column in outlier_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Column": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
# Create a dictionary to store the potential outlier records
# for each numerical column.

potential_outliers = {}

for column in outlier_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    potential_outliers[column] = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

# Display the potential outliers for each column.

for column, records in potential_outliers.items():
    print(f"\n--- Potential outliers in {column} ---")
    print(records[[column]].to_string(index=False))
```

```python
# Display the complete records that have potential
# Total_Sales outliers.

total_sales_outliers = potential_outliers["Total_Sales"]

total_sales_outliers
```
